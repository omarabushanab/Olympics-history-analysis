# Key Findings & Data Limitations

## 1. Dataset Structure

- Each record in `athlete_events.csv` represents an athlete's participation in an Olympic event.
- The dataset is sorted alphabetically by athletes' last names.
- The dataset covers the period from 1896 to 2016, including both Summer and Winter Olympic Games.

## 2. Athlete Identification

- `ID` is a unique identifier for each athlete, not for each participation record.
- The number of unique IDs is greater than the number of unique names, indicating that names alone are not reliable identifiers.
- An athlete can have multiple records because they may participate in multiple events or Olympic editions.

## 3. Historical Coverage

- Older Olympic records have more missing Height and Weight measurements.
- This may reflect limitations in historical data collection.
- The dataset includes historical Olympic participation records, including unusual cases that require further investigation.

## 4. Missing Data

The following columns contain missing values:

| Column | Observation |
|---|---|
| Age | Some athlete ages are unavailable. |
| Height | Many measurements, particularly in older records, are missing. |
| Weight | Many measurements, particularly in older records, are missing. |
| Medal | NaN indicates that the athlete did not win a medal. |

Missing measurements should not automatically result in removing participation records.

## 5. Duplicate Records

- The dataset contains 1,385 exact duplicate records.
- There are 1,480 duplicate athlete-event combinations.
- These records require further investigation before deciding whether any should be removed.

## 6. Country and Team Representation

- `Team` is not necessarily equivalent to `NOC`.
- For example, the USA has more records under different team names than under the single name "United States".
- `noc_regions.csv` provides the relationship between NOC codes and their corresponding country or region names.
- The NOC lookup dataset contains missing region names for the codes ROT, TUV, and UNK.
- The athlete dataset contains one NOC code (`SGP`) that is absent from the lookup dataset.
- `SGP` represents Singapore, so this is a missing lookup entry, not an invalid athlete record.
- The mapping should be added during data cleaning.

## 7. Unusual Observations

- John Quincy Adams appears with an age of 97, despite the historical circumstances suggesting posthumous participation.
- Winslow Homer also appears in the dataset.
- These observations require further verification rather than automatic removal.

## 8. Data Quality Considerations

Based on the initial investigation, the following issues require attention during data preparation:

1. Investigating and handling exact duplicate records.
2. Understanding duplicate athlete-event combinations.
3. Preserving legitimate missing values.
4. Investigating unusual numerical observations.
5. Validating NOC codes and country mappings.
6. Maintaining the distinction between athletes, participation records, teams, and countries.

**Conclusion:** The dataset provides extensive historical Olympic information, but its structure and historical limitations must be considered when performing analysis. Cleaning decisions should be based on the meaning of each variable and the requirements of the analysis.
